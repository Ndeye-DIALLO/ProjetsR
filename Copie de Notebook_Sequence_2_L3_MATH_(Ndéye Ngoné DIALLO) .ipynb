### 3.1 Histogramme — `hist()`
Pour visualiser la **distribution** d'une variable quantitative.

```r
hist(etudiants$note,
     main = "Distribution des notes",
     xlab = "Note / 20",
     ylab = "Effectif",
     col = "steelblue",
     breaks = 10)
```

### 3.2 Boîte à moustaches — `boxplot()`
Pour comparer une variable quantitative **entre plusieurs groupes**.

```r
boxplot(note ~ groupe, data = etudiants,
        main = "Notes par groupe de TD",
        xlab = "Groupe", ylab = "Note",
        col = c("lightblue","lightgreen","lightpink"))
```

### 3.3 Nuage de points — `plot()`
Pour visualiser la relation entre **deux variables quantitatives**.

```r
plot(etudiants$heures_revision, etudiants$note,
     main = "Note en fonction des heures de révision",
     xlab = "Heures de révision", ylab = "Note",
     pch = 19, col = "darkgreen")

# on peut ajouter une droite de régression
abline(lm(note ~ heures_revision, data = etudiants), col = "red", lwd = 2)
```

### 3.4 Diagramme en barres — `barplot()`
Pour visualiser les **effectifs** d'une variable qualitative. On utilise d'abord `table()`.

```r
effectifs_filiere <- table(etudiants$filiere)
effectifs_filiere

barplot(effectifs_filiere,
        main = "Répartition des étudiants par filière",
        xlab = "Filière", ylab = "Effectif",
        col = c("coral","gold","lightseagreen"))
```

### Exercice 3.4 (Barplot)

### 3.5 Diagramme circulaire — `pie()`
À utiliser avec parcimonie : seulement pour peu de catégories (2 à 5).

```r
effectifs_genre <- table(etudiants$genre)
pie(effectifs_genre,
    main = "Répartition par genre",
    col = c("pink","lightblue"),
    labels = paste(names(effectifs_genre), ":", effectifs_genre))
```

## 4. Quel graphique choisir ?

| Situation | Graphique |
|---|---|
| 1 variable quantitative → distribution | `hist()` |
| 1 variable quantitative comparée entre groupes | `boxplot()` |
| 2 variables quantitatives → relation | `plot()` |
| 1 variable qualitative → effectifs | `barplot()` |
| 1 variable qualitative → proportions (peu de catégories) | `pie()` |
